### Step 2
Initialize a tracker (e.g. KCF).

```python
import cv2
import os
from tqdm import tqdm
from matplotlib import pyplot as plt
```

```python

#=== Tracker type selection ===
trackerType = 'CSRT'

if trackerType == 'MIL':
    tracker1 = cv2.TrackerMIL_create()
    tracker2 = cv2.TrackerMIL_create()

elif trackerType == 'KCF':
    tracker1 = cv2.TrackerKCF_create()
    tracker2 = cv2.TrackerKCF_create()

elif trackerType == "CSRT":
    tracker1 = cv2.TrackerCSRT_create()
    tracker2 = cv2.TrackerCSRT_create()
```

```python
# window coodrinates for detected object
#aduld person
x11, y11 = 1190, 350
x12, y12 = 1320, 560
width1 = x12 - x11
height1 = y12 - y11

#children person
x21, y21 = 1093, 418
x22, y22 = 1151, 547
width2 = x22 - x21
height2 = y22 - y21
```

```python
# === Load video ===
inputVideo = cv2.VideoCapture('data/video.mp4')
inputWidth  = int(inputVideo.get(cv2.CAP_PROP_FRAME_WIDTH))
inputHeight = int(inputVideo.get(cv2.CAP_PROP_FRAME_HEIGHT))
print('Video resolution: {}x{}'.format(inputWidth, inputHeight))

# === Prepare output video ===
#Directory containing the output video
video_dir = 'video_output'
# === Create directory for saving frames ===
os.makedirs(video_dir, exist_ok=True)
# Output video
output_video_path = os.path.join(video_dir, f'final_output_{trackerType}.avi')
#outputfileName = f'output_{trackerType}.avi'
outputVideo = cv2.VideoWriter(output_video_path, cv2.VideoWriter_fourcc(*"MJPG"), 20.0, (inputWidth, inputHeight))

#Directory containing the frames
frames_dir = f'frames_{trackerType}'

# === Create directory for saving frames ===
os.makedirs(frames_dir, exist_ok=True)
```

```python
# === Initialize trackers ===
ret, frame = inputVideo.read()
tracker1.init(frame, (x11, y11, width1, height1))
tracker2.init(frame, (x21, y21, width2, height2))
```

### Step 3 + Step 4
Run the tracker on the video and the selected object. Run the tracker for around 10-15 frames.
For each frame, print the bounding box on the image and save it.

```python
# === Function to draw bounding box ===
def printBox(tracker, frame):
    ok, bbox = tracker.update(frame)
    print(ok, bbox)
    
    if ok:
        x1, y1 = int(bbox[0]), int(bbox[1])
        width, height = int(bbox[2]), int(bbox[3])
        cv2.rectangle(frame, (x1, y1), (x1 + width, y1 + height), (0, 255, 0), 2)
    else:
        cv2.putText(frame, "Tracking failure", (50, 80), cv2.FONT_HERSHEY_SIMPLEX, 0.75, (0, 0, 255), 2)
    
    return frame

#count total frames for tqdm    
total_frames = int(inputVideo.get(cv2.CAP_PROP_FRAME_COUNT))

# === Process video frames ===
frame_number = 0

with tqdm(total=total_frames, desc="Tracking Progress", unit="frame") as pbar:
    while inputVideo.isOpened():
        ret, frame = inputVideo.read()
        
        if ret:
            frame = printBox(tracker1, frame)
            frame = printBox(tracker2, frame)

            filename = os.path.join(frames_dir, f"frame_{frame_number:04d}.jpg")
            cv2.imwrite(filename, frame)
            outputVideo.write(frame)

            frame_number += 1
            pbar.update(1)
        else:
            break

# === Release resources ===
inputVideo.release()
outputVideo.release()
print("Tracking complete. Frames and video saved.")
```
