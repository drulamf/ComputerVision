### Step 0

Run the necessary imports.

```python
import dlib
import cv2
import numpy as np
from matplotlib import pyplot as plt
import random
```

### Step 1

Load an image (any image that contains faces).

```python
img = cv2.imread('data/event.jpg')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)  # Convert to RGB for display
gray = cv2.cvtColor(img, cv2.COLOR_RGB2GRAY)
plt.imshow(img)
```

### Step 2

Load the dlib face predictor.

```python
# Initialize Dlib's face detector
detector = dlib.get_frontal_face_detector()
```

### Step 3

Run the predictor on your image.

```python
# Detect faces in the image
faces = detector(gray)

# Print the number of faces found
print(f"Number of faces detected: {len(faces)}")
```

### Step 4

Draw bounding boxes around the detected faces and plot the image. Use different colour for each face.

```python
#Generate a random color for each face
colors = [
    (random.randint(0, 255), random.randint(0, 255), random.randint(0, 255))
    for _ in range(len(faces))
]

# Draw bounding boxes
for i, face in enumerate(faces):
    x, y, w, h = face.left(), face.top(), face.width(), face.height()
    cv2.rectangle(img_rgb, (x, y), (x + w, y + h), colors[i], 2)

# Display the result
plt.imshow(img_rgb)
plt.axis('off')
plt.title("Detected Faces with Unique Colors")
plt.show()
```
